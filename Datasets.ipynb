```python
from pathlib import Path
from pypdf import PdfWriter
import random
random.seed(42)
```

```python
def batch_concat_pdfs(input_dir: str, output_dir: str, batch_size: int = 3):
    """
    Groups individual PDFs from input_dir into concatenated batch PDFs using PdfWriter.
    """
    input_path = Path(input_dir)
    output_path = Path(output_dir)
    # output_path.mkdir(parents=True, exist_ok=True)

    # Collect all single resume PDFs
    pdf_files = list(input_path.glob("*.pdf"))
    random.shuffle(pdf_files)
    print('num pdfs: ', len(pdf_files))

    if not pdf_files:
        print(f"No PDF files found in {input_dir}")
        return

    # Process files in chunks of `batch_size`
    for i in range(0, len(pdf_files), batch_size):
        batch = pdf_files[i : i + batch_size]
        writer = PdfWriter()

        for pdf_file in batch:
            # append() adds all pages from the given PDF path
            writer.append(pdf_file)

        batch_idx = (i // batch_size) + 1
        out_file = output_path / f"batch_resumes_{batch_idx}.pdf"
        
        # Write out the concatenated PDF
        with open(out_file, "wb") as f_out:
            writer.write(f_out)
            
        writer.close()
        print(f"Created '{out_file.name}' containing {len(batch)} resumes.")
```

```python
input_dir = "./resume_pdfs"
output_dir = "./concatenated_batches"
batch_size=5
```

```python
batch_concat_pdfs(input_dir, output_dir, batch_size)
```
